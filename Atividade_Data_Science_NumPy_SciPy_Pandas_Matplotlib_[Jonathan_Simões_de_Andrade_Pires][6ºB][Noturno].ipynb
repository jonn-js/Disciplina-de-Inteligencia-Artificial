21. Busca Binária

Explicação: Criei uma função que divide a lista ordenada ao meio até encontrar o índice do número procurado.

```python
def busca_binaria(lista, numero):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == numero:
            return meio
        elif lista[meio] < numero:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1


lista = [1, 3, 5, 7, 9, 11]
numero = int(input("Digite o número que deseja procurar: "))

resultado = busca_binaria(lista, numero)

if resultado != -1:
    print("Encontrado na posição:", resultado)
else:
    print("Número não encontrado.")
```

22. Criptografia de Cifra de César

Explicação: Combinei ord() e chr() com a fórmula de deslocamento e o resto da divisão para alterar as posições das letras no alfabeto.

```python
def cesar(texto, deslocamento):
    resultado = ""

    for letra in texto:
        if letra.isalpha():
            inicio = ord("A") if letra.isupper() else ord("a")
            nova_letra = chr((ord(letra) - inicio + deslocamento) % 26 + inicio)
            resultado += nova_letra
        else:
            resultado += letra

    return resultado


mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:", cesar(mensagem, deslocamento))
```

23. Descriptografar Cifra de César

Explicação: Utilizei a lógica da Cifra de César, mas subtraí o valor do deslocamento para reverter a mensagem para o texto original.

```python
def descriptografar(texto, deslocamento):
    resultado = ""

    for letra in texto:
        if letra.isalpha():
            inicio = ord("A") if letra.isupper() else ord("a")
            nova_letra = chr((ord(letra) - inicio - deslocamento) % 26 + inicio)
            resultado += nova_letra
        else:
            resultado += letra

    return resultado


mensagem = input("Digite a mensagem cifrada: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem original:", descriptografar(mensagem, deslocamento))
```

24. Validação de CPF

Explicação: Removi os caracteres de pontuação com .replace() e executei o cálculo ponderado dos dois dígitos verificadores para validar o número.

```python
def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")

    if len(cpf) != 11 or not cpf.isdigit():
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11
    digito1 = 0 if resto < 2 else 11 - resto

    if int(cpf[9]) != digito1:
        return False

    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11
    digito2 = 0 if resto < 2 else 11 - resto

    return int(cpf[10]) == digito2


cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

30. Jogo da Forca

Explicação: Representei a palavra oculta em uma lista de "_" e usei o while para substituir pelas letras acerta a cada palpite.

```python
palavra = "alecrim"
descobertas = ["_"] * len(palavra)
tentativas = 6
letras_tentadas = []

while tentativas > 0 and "_" in descobertas:
    print("\nPalavra:", " ".join(descobertas))
    print("Tentativas restantes:", tentativas)

    letra = input("Digite uma letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Digite apenas uma letra.")
        continue

    if letra in letras_tentadas:
        print("Você já tentou essa letra.")
        continue

    letras_tentadas.append(letra)

    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                descobertas[i] = letra
        print("Acertou!")
    else:
        tentativas -= 1
        print("Errou!")

if "_" not in descobertas:
    print("\nVocê venceu! Palavra:", palavra)
else:
    print("\nVocê perdeu! Palavra:", palavra)
```

31. Ordenação por Bolha

Explicação: Montei o algoritmo Bubble Sort com dois for para comparar pares vizinhos e trocar as posições quando estiverem fora de ordem.

```python
numeros = [5, 3, 8, 1, 2]

for i in range(len(numeros)):
    for j in range(len(numeros) - 1 - i):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(numeros)
```

32. Jogo da Velha

Explicação: Criei uma lista de 9 posições para o tabuleiro e escrevi a função venceu() para checar as possíveis combinações de vitória.

```python
tabuleiro = [" "] * 9

def mostrar():
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("--+---+--")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("--+---+--")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])

def venceu(jogador):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    return any(
        all(tabuleiro[posicao] == jogador for posicao in combinacao)
        for combinacao in combinacoes
    )

jogador = "X"

for rodada in range(9):
    mostrar()

    while True:
        try:
            posicao = int(input(f"Jogador {jogador}, escolha 1 a 9: ")) - 1
            if 0 <= posicao <= 8 and tabuleiro[posicao] == " ":
                break
            print("Posição inválida ou ocupada.")
        except ValueError:
            print("Digite um número de 1 a 9.")

    tabuleiro[posicao] = jogador

    if venceu(jogador):
        mostrar()
        print("Jogador", jogador, "venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar()
    print("Empate!")
```

46. Calcular Média e Desvio Padrão

Explicação: Utilizei a média simples e calculei o desvio padrão aplicando a raiz quadrada com math.sqrt() sobre a soma dos desvios ao quadrado.

```python
import math

numeros = [10, 20, 30, 40, 50]

media = sum(numeros) / len(numeros)

soma = 0
for numero in numeros:
    soma += (numero - media) ** 2

desvio = math.sqrt(soma / len(numeros))

print("Média:", media)
print("Desvio padrão:", desvio)
```

48. Ordenação por Inserção

Explicação: Implementei o Insertion Sort com um while para comparar e puxar o número para a esquerda até encaixar na posição correta.

```python
numeros = [5, 2, 4, 6, 1, 3]

for i in range(1, len(numeros)):
    chave = numeros[i]
    j = i - 1

    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = chave

print(numeros)
```

49. Soma dos Dígitos

Explicação: Usei o operador % 10 para extrair o último dígito do número e // 10 para remover, somando os valores em um while.

```python
numero = int(input("Digite um número inteiro: "))

numero = abs(numero)
soma = 0

while numero > 0:
    digito = numero % 10
    soma += digito
    numero //= 10

print("Soma dos dígitos:", soma)
```
