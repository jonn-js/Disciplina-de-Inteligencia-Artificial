19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

a = np.random.randint(1, 10, 20)
b = np.random.randint(1, 10, 20)

produto = np.dot(a, b)

angulo = np.arccos(produto / (np.linalg.norm(a) * np.linalg.norm(b)))

print("Ângulo em radianos:", angulo)
print("Ângulo em graus:", np.degrees(angulo))
```

2 - Exercícios de SciPy

1. Calcule a integral da função f(x)=x 2 sin(x) no intervalo de 0 a π.

```python
from scipy.integrate import quad
import numpy as np

funcao = lambda x: x**2 * np.sin(x)

resultado = quad(funcao, 0, np.pi)

print(resultado[0])
```

2. Resolva a equação diferencial dx dy =x−y, com y(0)=1, no intervalo de 0 a 10.

```python
from scipy.integrate import solve_ivp

def funcao(x, y):
    return x - y

resultado = solve_ivp(funcao, [0, 10], [1], t_eval=range(11))

print(resultado.y[0])
```

3. Encontre as raízes da função f(x)=x 4 −3x 3 +2 utilizando SciPy.

```python
import numpy as np

coeficientes = [1, -3, 0, 0, 2]

raizes = np.roots(coeficientes)

print(raizes)
```

4. Use a função de otimização do SciPy para maximizar f(x)=−x 2 +4x+1.

```python
from scipy.optimize import minimize

funcao = lambda x: -( -x[0]**2 + 4*x[0] + 1 )

resultado = minimize(funcao, [0])

print("Máximo em x =", resultado.x[0])
```

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 100)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)

transformada = fft(sinal)

print(transformada)
```

10. Calcule a integral dupla da função f(x,y)=xy+x 2 no intervalo [0, 2] para x e [0, 1] para y.

```python
from scipy.integrate import dblquad

funcao = lambda y, x: x*y + x**2

resultado = dblquad(funcao, 0, 2, lambda x: 0, lambda x: 1)

print(resultado[0])
```
